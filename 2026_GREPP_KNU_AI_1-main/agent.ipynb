```python
from langchain.agents import create_agent
from langchain_core.tools import tool

from langchain_groq import ChatGroq
from dotenv import load_dotenv

load_dotenv()

@tool
def recommend_recipe(ingredient: str) -> str:
    """사용자가 가진 재료로 만드는 요리 추천"""

    return f"""
가진 재료 : {ingredient}

추천 레시피 :
1. 요리 이름 : 오무라이스
2. 추가 재료 : 소금, 식용유
3. 요리 과정 :
   1. 달걀을 깨서 섞는다.
   2. 채소를 먼저 팬에 볶는다.
   3. 달걀을 익힌다.
4. 조리 시간 : 약 20분
"""


model = ChatGroq(
    model="openai/gpt-oss-20b",
    temperature=0.7,
    api_key="키값"
)

agent = create_agent(
    model=model,
    tools=[recommend_recipe]
)

response = agent.invoke({
    "messages": [
        {
            "role": "user",
            "content": "달걀을 사용한 요리 추천"
        }
    ]
})

print(response['messages'][-1].content)

```
